```python
missing = df['Applicant_ID'].isnull()

start = int(df['Applicant_ID'].max()) + 1

df.loc[missing, 'Applicant_ID'] = range(
    start,
    start + missing.sum()
)
```

```python
cols = ['Applicant_Income', 'Coapplicant_Income', 'Age','Dependents','Credit_Score','Existing_Loans','DTI_Ratio','Savings','Collateral_Value','Loan_Amount','Loan_Term']
df[cols] = df[cols].fillna(df[cols].mean())
```

```python
df['Employment_Status'] = df['Employment_Status'].fillna('Unknown')
df['Marital_Status'] = df['Marital_Status'].fillna('Unknown')
df['Loan_Purpose'] = df['Loan_Purpose'].fillna('Unknown')
df['Property_Area'] = df['Property_Area'].fillna('Unknown')
df['Education_Level'] = df['Education_Level'].fillna('Unknown')
df['Gender'] = df['Gender'].fillna('Unknown')
df['Employer_Category'] = df['Employer_Category'].fillna('Unkown')
```

```python
df.info()
```

```python
df = df.dropna(subset=['Loan_Approved'])
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    df.drop(columns=['Loan_Approved']),
    df['Loan_Approved'],
    test_size=0.2,
    random_state=42
)
```

```python
x_train.sample(5)
```

```python
x_train['Employment_Status'].value_counts()
```

```python
mapping = {
    'Salaried': 1,
    'Contract': 1,
    'Self-employed': 1,
    'Unemployed': 0,
    'Unknown': 0
}

x_train['Employment_Status'] = x_train['Employment_Status'].map(mapping)
x_test['Employment_Status'] = x_test['Employment_Status'].map(mapping)
```

```python
x_train['Marital_Status'].value_counts()
```

```python
mapping = {
    'Single': 1,
    'Married': 0,
    'Unknown': 0
}

x_train['Marital_Status'] = x_train['Marital_Status'].map(mapping)
x_test['Marital_Status'] = x_test['Marital_Status'].map(mapping)
```

```python
x_train.head()
```

```python
x_train['Loan_Purpose'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

encoded_train = encoder.fit_transform(x_train[['Loan_Purpose']])
encoded_test = encoder.transform(x_test[['Loan_Purpose']])

x_train[encoder.get_feature_names_out()] = encoded_train
x_test[encoder.get_feature_names_out()] = encoded_test

x_train.drop(columns=['Loan_Purpose'], inplace=True)
x_test.drop(columns=['Loan_Purpose'], inplace=True)
```

```python
df['Education_Level'].value_counts()
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

train = encoder.fit_transform(x_train[['Property_Area']])
test = encoder.transform(x_test[['Property_Area']])

x_train[encoder.get_feature_names_out()] = train
x_test[encoder.get_feature_names_out()] = test

x_train.drop(columns='Property_Area', inplace=True)
x_test.drop(columns='Property_Area', inplace=True)
```

```python
mapping = {
    'Graduate': 1,
    'Not Graduate': 0,
    'Unknown': 0
}

x_train['Education_Level'] = x_train['Education_Level'].map(mapping)
x_test['Education_Level'] = x_test['Education_Level'].map(mapping)
```

```python
mapping = {
    'Male': 1,
    'Female': 0,
    'Unknown': 0
}

x_train['Gender'] = x_train['Gender'].map(mapping)
x_test['Gender'] = x_test['Gender'].map(mapping)
```

```python
df['Employer_Category'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

train = encoder.fit_transform(x_train[['Employer_Category']])
test = encoder.transform(x_test[['Employer_Category']])

x_train[encoder.get_feature_names_out()] = train
x_test[encoder.get_feature_names_out()] = test

x_train.drop(columns='Employer_Category', inplace=True)
x_test.drop(columns='Employer_Category', inplace=True)
```

```python
y_train.value_counts()
```

```python
mapping = {
    'Yes': 1,
    'No': 0
}

y_train = y_train.map(mapping)
y_test = y_test.map(mapping)
```

```python
y_train.head()
```

```python
x_train.info()
```

```python
x_train.describe()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled[:1]
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

model = KNeighborsClassifier()

params = {
    'n_neighbors': [3, 5, 7, 9, 11, 13, 15, 17, 19]
}
                      
grid = GridSearchCV(
    estimator=model,
    param_grid=params,
    cv=5
)


grid.fit(x_train_scaled, y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

y_pred = grid.predict(x_test_scaled)
```

```python
print('accuracy_score: ', accuracy_score(y_test, y_pred))
print('precision_score: ', precision_score(y_test, y_pred))
print('recall_score: ', recall_score(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train_scaled, y_train)
```

```python
y_pred_lr = lr.predict(x_test_scaled)
```

```python
print('accuracy_score: ', accuracy_score(y_test, y_pred_lr))
print('precision_score: ', precision_score(y_test, y_pred_lr))
print('recall_score: ', recall_score(y_test, y_pred_lr))
```

```python
print(confusion_matrix(y_test, y_pred_lr))
```
